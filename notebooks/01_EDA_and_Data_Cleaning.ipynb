```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All required libraries imported successfully!")
```

```python
file_path = "../data/raw_transactions.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())
print("Negative prices:", (df["Price"] < 0).sum())
print("Zero prices:", (df["Price"] == 0).sum())
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")

df["Revenue"] = df["Quantity"] * df["Price"]

print("InvoiceDate converted successfully!")
print("Invalid dates:", df["InvoiceDate"].isna().sum())

print("\nDate range:")
print(df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())

print("\nRevenue summary:")
print(df["Revenue"].describe())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Remove transactions without Customer ID
df = df.dropna(subset=["Customer ID"])

# Remove zero-price transactions
df = df[df["Price"] > 0]

# Keep only positive quantities for customer revenue analysis
df = df[df["Quantity"] > 0]

print("Cleaning completed!")
print("Cleaned shape:", df.shape)
print("Missing Customer IDs:", df["Customer ID"].isna().sum())
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero prices:", (df["Price"] == 0).sum())
```

```python
print("Cleaned dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nUnique customers:", df["Customer ID"].nunique())
print("Unique invoices:", df["Invoice"].nunique())
print("Unique countries:", df["Country"].nunique())

print("\nCleaned data preview:")
display(df.head())
```

```python
cleaned_path = "../data/cleaned_transactions.csv"

df.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print("File:", cleaned_path)
print("Rows:", len(df))
```
